```python
def add(a,b):
  return a+b

#my_addition_operator = add
#print(my_addition_operator(2,3))

def sub(a,b):
  return a-b

def mult(a,b):
  return a*b

def div(a,b):
  return a/b

operations = {
    "+": add,
    "-": sub,
    "*": mult,
    "/": div
}


def calculator():
  should_accumulate = True
  num1 = float(input("Enter the value of first num: "))

  while should_accumulate:
    for symbols in operations :
      print(symbols)

    operation_symbol = input("Enter the symbol of operation you want to perform ")

    num2 = float(input("Enter the value of second num: "))

    #print(f"{operations[operation_symbol](num1 , num2)}")
    answer = operations[operation_symbol](num1 , num2)
    print(f"{num1} {operation_symbol} {num2} = {answer} ")

    choice = input(f"Enter 'y' if you want to continue with the {answer} otherwise type 'n' for stop calculations: ").lower()

    if choice == "y":
      num1= answer
    else:
      should_accumulate = False
      print("\n"  *20)
      # Removed the recursive call to calculator() here.
      # The function will now simply terminate when should_accumulate is False.

# Initial call to start the calculator.
# If the user chooses 'n' inside the function, this initial call will eventually complete.
calculator()
```
