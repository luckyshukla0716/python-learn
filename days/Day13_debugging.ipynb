```python
#Fixing Errors and Watching for Red Underlines
#solution debugging:
try:
  age = int(input("What is your age? " ))
except ValueError:
  print("You have typed wrong, we are expecting a number here :")
  age = int(input("What is your age? " ))

if age > 18:
  print(f"You can drive at {age}.")
```
