```python
import random
import math

def deal_cards():
  cards = [11,2,3,4,5,6,7,8,9,10,10,10,10] #here A = 11, J=10, K=10, Q=10
  card = random.choice(cards)
  return card

def calculate_score(cards):
  if sum(cards)==21 and len(cards) ==2:
    return 0
  if 11 in cards and sum(cards) > 21:
    cards.remove(11)
    cards.append(1)

  return sum(cards)

def compare_score(u_score , c_score):
  # Check for Blackjacks first
  if u_score == 0 and c_score == 0:
    return "Draw (Both Blackjacks)"
  elif u_score == 0:
    return "You win with a Blackjack!"
  elif c_score == 0:
    return "Computer wins with a Blackjack!"

  # Check for busts
  if u_score > 21 and c_score > 21:
    return "You both went over 21. Computer wins!"
  elif u_score > 21:
    return "You went over 21. You lose!"
  elif c_score > 21:
    return "Computer went over 21. You win!"

  # Compare scores if no blackjacks or busts
  if u_score == c_score:
    return "Draw!"
  elif u_score > c_score:
    return "You win!"
  else:
    return "Computer wins!"

def play_game():
    user_cards = []
    computer_cards = []
    computer_score = -1
    user_score = -1
    is_game_over = False

    for _ in range(2):
      user_cards.append(deal_cards())
      computer_cards.append(deal_cards())

    while not is_game_over:

        user_score = calculate_score(user_cards)
        computer_score = calculate_score(computer_cards)

        print(f"\n--- Your Turn ---\n")
        print(f"🃏 Your cards: {user_cards},  Current score: {user_score} ")
        print(f"🤖 Computer's first card: {computer_cards[0]}")
        print(f"\n-------------------\n")

        if user_score == 0 or computer_score == 0 or user_score > 21 :
          is_game_over = True
          # End the loop immediately if game is over (e.g., user busts)
          break
        else :
          user_should_deal = input("type 'y' if you want to continue otherwise type 'n' ").lower()
          if user_should_deal == "y":
            user_cards.append(deal_cards())
          else:
            is_game_over = True

    # Only run computer's turn if user hasn't busted and game is not already over
    if not is_game_over:
        while computer_score !=0 and  computer_score < 17:
          computer_cards.append(deal_cards())
          computer_score = calculate_score(computer_cards)

    print(f"🃏 Your final hand : {user_cards} , final score : {user_score} ")
    print(f"🤖 Computer's final hand : {computer_cards} , final score : {computer_score} ")
    print(compare_score(user_score , computer_score))


while input("Do you want to play a game of Blackjack? Type 'y' or 'n': ") == 'y':
    play_game()
```
