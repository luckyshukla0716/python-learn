```python
def outer_function(a, b):
    def inner_function(c, d):
        return c + d
    return inner_function(a, b)

result = outer_function(5, 10)
print(result)

# Explanation:
# This code demonstrates nested functions. 'inner_function' is defined within 'outer_function'.
# 'outer_function' takes two arguments (a, b).
# It then defines 'inner_function' which takes two arguments (c, d) and returns their sum.
# 'outer_function' calls 'inner_function' with its own arguments (a, b) and returns the result of that call.
# In this specific case, outer_function(5, 10) leads to inner_function(5, 10) being called,
# which returns 15. This value is assigned to 'result' and then printed.
```
